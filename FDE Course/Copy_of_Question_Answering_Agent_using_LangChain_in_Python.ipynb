## Importing Libraries

```python
import os

from google.colab import userdata
from IPython.display import Markdown, display
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
```

## Loading the OPENAI API Key

```python
os.environ["GOOGLE_API_KEY"] = userdata.get("GOOGLE_API_KEY")
```

## Declaring the OPENAI Model and asking a simple question

```python
model = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0.1)#temperature is used to control the randomness of the output. A higher value corresponds to higher randomness
text = "What is paranormal activity,keep it simple?" #prompt
print(model.invoke(text).content)
```

## Creating a Multi-Template for grouping multiple questions in one-single prompt

```python
mu_template = """Answer the following questions ONLY one at a time.
Questions:
{ques}
Answers:
"""
```

## Passing the template in our prompt and using text-davinci-003 OPENAI model

```python
long_prompt = PromptTemplate(template=mu_template, input_variables=["ques"])
davinci = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

## Building a LLMChain

```python
llm_chain = long_prompt | davinci

ques = (
    "Mention one of the famous paranormal activity in 1994?\n" +
    "What can be a solution for such paranormal activities" +
    "some theory that suggest such paranormal activites are just illustions and not exacely true, explain that point of view?"
)
response = llm_chain.invoke({"ques": ques})
display(Markdown(response.text))
```

## Running the chain and checking the results

```python
print(llm_chain.run(ques))
```
